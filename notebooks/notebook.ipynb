```python
from job_agent.model import load_model
from job_agent.tools import send_mail
from job_agent.prompts import SYSTEM_PROMPT
from langchain.messages import HumanMessage, SystemMessage

print('System prompt', SYSTEM_PROMPT)

tools = [send_mail]
llm = load_model()
input_message = [
    SystemMessage(content=SYSTEM_PROMPT),
    HumanMessage(content="hallow")
]

response = llm.invoke(input_message)
print(response.content)
print(response.tool_calls)


llm = llm.bind_tools([send_mail])
response = llm.invoke(input_message)
print(response.content)
print(response.tool_calls)

```

```python
import urllib.request
import urllib.error

def is_ollama_running(url="http://localhost:11434"):
    try:
        # Send a request to the default Ollama local server
        response = urllib.request.urlopen(url, timeout=2)
        if response.status == 200:
            return True
    except urllib.error.URLError:
        # Server is not running or port is closed
        return False
```
